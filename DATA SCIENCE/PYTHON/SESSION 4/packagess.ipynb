```python
def seriesSum(a, n):
    if n<=0:
        return 0
    else:
        return pow(a, n) + seriesSum(a, n-1)

seriesSum (4,6)
```

```python
def factoria(n):
    if n <0:
        return "Factoria not defines"
    elif n==0 or n==1:
        return 1
    else:
        return n * factoria (n-1)

factoria (4)
```

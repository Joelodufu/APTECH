```python
import re
text = "hello i 34576 have a cat _in my \nhouse, it likes eating rat"
expression = r"^\w{3}"
expression2 = r"[a-zA-Z0-9_]{3}"

result = re.findall(expression, text, re.M)
print(result)
```

```python
text = "Rose is red in color. Read newspaper daily. Religion"
expression = r"r.*? "
result = re.findall(expression, text, re.I)
print(result)
```

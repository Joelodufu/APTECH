```python
def calculateInterest(sellPrice, costPrice):
    profit=sellPrice - costPrice
    if (profit<0):
        print("We made a lose of", profit*(-1))
    else:
        print("We made a profit of", profit)

calculateInterest(200000, 3500000)
```

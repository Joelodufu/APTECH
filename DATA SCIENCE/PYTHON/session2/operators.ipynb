```python
a = 5
b = 10

# Arithmetic Operators
print(a + b)          # Output: 15
print( b - a)       # Output: 5
print(a * b)         # Output: 50
print(b / a)         # Output: 2.0
print(b % a)         # Output: 0
print(a ** 2)        # Output: 25
```

```python
x = 7
y = 10

# Comparison Operators
print(x == y)        # Output: False
print(x != y)        # Output: True
print(x < y)         # Output: True
print(x > y)         # Output: False
print(x <= y)       # Output: True
print(x >= y)       # Output: False
```

```python
age = 20
country = "Nigeria"
# Logical Operators
print(age > 18 and country == "Nigeria")  # Output: True
print(age < 18 or country == "Nigeria")   # Output: True
print(not (age < 18))                      # Output: True
```

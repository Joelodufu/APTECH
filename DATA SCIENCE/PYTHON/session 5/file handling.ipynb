```python
file = open("info.txt", "w")
file.write("Hello we are writing into this file")
file.close()
```

```python
file = open("info.txt", "a")
file.write("\nI am in the new line")
file.close()
```

```python
with open("message.txt", "w") as file:
    file.write("Hello this is the message from the King")
```

```python
with open("message.txt", "a") as file:
    file.write("\nNext Line Message")
```

```python
with open("message.txt", "a+") as file:
    print("Before: ", file.read())
    file.seek(0)
    file.write("\nfgfgfg text.")
```

```python
with open("message.txt", "r") as file:
    content = file.readlines()
    print(content)
```

```python
lines = [
    "Joel Odufu\n",
    "Smart Eremba\n",
    "Jude Samuel\n",
    "Praise Peters\n"
]

with open("message.txt", "w") as file:
    file.writelines(lines)
```

```python
with open("message.txt", "r+") as file:
    file.truncate(10)
    print(file.read())
```

```python
with open("message.txt", "r") as file:
    print (file.tell())
    file.read(10)
    print(file.tell())
```

```python
with open("message.txt", "r") as file:
    # file.read(3)
    file.seek(7,0)
    print(file.read())
```

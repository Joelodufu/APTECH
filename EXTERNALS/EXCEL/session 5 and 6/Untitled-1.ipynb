```python
def almighty(a,b,c):
    descriminant = b**2 -(4*a*c)
    print(descriminant)
    # 2 distinct roots 
    if(descriminant>0):
        x1 =( -b+(descriminant**0.5))/2*a
        x2 =( -b-(descriminant**0.5))/2*a
        return(x1,x2)
    # 1 distict root
    elif(descriminant==0):
        x =( -b+(descriminant**0.5))/2*a
        return(x)
    # no real root
    else:
        return("No Real Roots")



almighty(1,-2,1)
```

```python
#SIMPLE INTEREST
#SI = P (PRINCIPAL: Investment/Borrowing) x R (RATE: Percentage in decimal.e.g 10%=0.1) x T (TIME: Time taken annually)
def simpleInterest (P, R,T):
    i= P*R*T
    return i
simpleInterest(10000, 0.15, 2)
```

```python
#COMPUND INTEREST
# A = The total amount of money you end up with (principal + all the interest earned).
# P = The starting money (principal)—like the seed you plant.
# r = The yearly interest rate as a decimal (e.g., 5% = 0.05). This is how much "growth" you get per year.
# n = How many times the interest gets added per year (e.g., 12 for monthly, 4 for quarterly, 1 for yearly).
# t = How many years you're letting it grow.
# The exponent (nt) = Total number of "compounding periods" (like mini-growth spurts).
```

```python
def compoundInterst(P, R, T,N=1 ):
    i = P*(1+(R/N))**(N*T)
    return i

compoundInterst(12000, 0.3, 12, 12)
```

```python
def averagScore(*scores):
    total=0
    for score in scores:
        total+=score
    average = total/len(scores)
    return average
averagScore(55,34,100,45,3444,2323)
```

```python
def MyMarks(name, **subject):
    total=0
    for sub in subject:
        subject_name = sub
        subject_mark=subject[sub]
        print(subject_name, "=", subject_mark)
        total+=subject_mark
    print("Total marks of ", name, " is = ", total)

MyMarks("Jude", English=20, Maths=25, Social_Scienc=45)
```
